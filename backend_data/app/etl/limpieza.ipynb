```python
df_products = pd.read_csv("data_raw/products.csv")
df_products.product_name.unique()
```

```python
print(f"Product_id únicos: {df_products['product_name'].nunique()}")
```

```python
# Resumen completo de duplicados
print(f"Total de registros: {len(df_products)}")
print(f"Product_id únicos: {df_products['product_id'].nunique()}")
print(f"Product_id duplicados: {df_products['product_id'].duplicated().sum()}")

# Mostrar los product_id que aparecen más de una vez
duplicados_detalle = df_products[df_products['product_id'].duplicated(keep=False)].sort_values('product_id')
if len(duplicados_detalle) > 0:
    print("\nRegistros duplicados:")
    print(duplicados_detalle)
```
